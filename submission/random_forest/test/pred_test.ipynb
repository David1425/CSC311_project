```python
import sys
import numpy as np
import pandas as pd
import pickle
sys.path.append('..')

from dataloader import DataLoader
from constants import PREDICTIONS_TO_LABELS

rf_path = '../safe/fitted_bow_rf_cmprsd_std.json'
count_vectorizer_path = '../count_vectorizer.pkl'
test_data_path = 'test_data_no_labels.csv'
vocab_path = 'vocab.csv'
columns_path = 'columns.csv'
json_dump = 'clean_test_data_no_labels.csv'
markus_data = 'markus_auto.csv'
labelled_test_data_path = '../../../data/test_data.csv'

with open(count_vectorizer_path, 'rb') as f:
    count_vectorizer = pickle.load(f)
```
