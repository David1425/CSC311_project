### 1. RF Test

```python
import sys
import numpy as np
sys.path.append('..')

from dataloader import DataLoader
from random_forest import RandomForestModel
from constants import PREDICTIONS_TO_LABELS


json_path = '../fitted_bow_rf_cmprsd_std.json'
test_data_path = 'test_data_no_labels.csv'
vocab_path = 'vocab.csv'
columns_path = 'columns.csv'
json_dump = 'clean_test_data_no_labels.csv'


dataloader = DataLoader()

# Extract vocabulary from training columns CSV
dataloader.extract_vocab_from_columns_csv(columns_path, vocab_path)
```

```python
# Create dataloader and load the vocabulary and expected columns
dataloader = DataLoader()
dataloader.load_vocab(vocab_path)
expected_columns = dataloader.load_column_names(columns_path)

# Process test data with expected columns
processed_data = dataloader.preprocess(test_data_path, expected_columns=expected_columns)
# processed_data.to_csv(json_dump)

# Make predictions
rf = RandomForestModel(json_path)
base_predictions = rf.predict(processed_data.to_numpy())

vectorized_map = np.vectorize(PREDICTIONS_TO_LABELS.get)
print(f"Cleaned predictions: {vectorized_map(base_predictions)}\n")
```
